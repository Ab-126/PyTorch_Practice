###Reshaping, stacking, squeezing and unsqueezing tensors

```python
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Add an extra dimension
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

```python
# Change the view -> Changing view changes the original tensor, bcz they share the same memory
z = x.view(1, 9)
z, z.shape
```

```python
z[:, 0] = 5
z, x
```

```python
x_stacked_1 = torch.stack([x, x], dim=0) # Vertical stacking
x_stacked_2 = torch.stack([x, x], dim=1) # Horizonal stacking
x_stacked_1, x_stacked_2
```

```python
x_reshaped, x_reshaped.shape
```

```python
squeezed = x_reshaped.squeeze()
squeezed, squeezed.shape
```

```python
squeezed.unsqueeze(dim=0), squeezed.unsqueeze(dim=0).shape
```

```python
#permute - rearranges the dimensions of a target tensor in a specified order
x_original = torch.rand(size=(224, 224, 3)) # [height, widht, color_channels]

x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0

print(f"Prev shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
```

###Reproduciability (trying to take random out of random)

```python
# Create 2 random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Let's make some random but reproducible tensors
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```

###Putting tensors (and models) on the GPU

```python
# Create a tensor (default on the CPU)
tensor = torch.tensor([1, 2, 3], device='cpu')

# Tensor not on GPU
print(tensor, tensor.device)
```

```python
# Move tensor to GPU (if available)
tensor_on_gpu = tensor.to(device)
tensor_on_gpu
```

```python
# Move tensors back to the cpu
# If tensor is on GPU, can't transform it to NumPy

tensor_back_on_cpu = tensor_on_gpu.cpu()
tensor_back_on_cpu.numpy()
```
